```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/Users/fuyuxuan/Downloads/testout_1.4.csv")
A = df.to_numpy(dtype=float)
A = (A + A.T) / 2.0

vals, vecs = np.linalg.eigh(A)
vals = np.clip(vals, 0.0, None)

B = vecs @ np.diag(vals) @ vecs.T
s = np.sqrt(np.diag(B))
C_psd = B / np.outer(s, s)

out = pd.DataFrame(C_psd, columns=df.columns)
print(out)
```
