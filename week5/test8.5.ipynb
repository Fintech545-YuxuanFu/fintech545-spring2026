```python
import numpy as np
import pandas as pd
from scipy.stats import t
```

```python
df = pd.read_csv("/Users/fuyuxuan/Downloads/test7_2.csv")
x = df.iloc[:, 0]    # get the only first column

# Fit the student-t distribution to estimate the degree of freedom
nu, loc, scale = t.fit(x)

alpha = 0.05

# Student t distribution critical values
t_alpha = t.ppf(alpha, nu)

# pdf at that critical value
pdf_val = t.pdf(t_alpha, nu)

# ES for Student t
es = loc + scale * (- (pdf_val / alpha) * ((nu + t_alpha**2) / (nu - 1)))

# ES as a positive loss
es_abs = - es

# Difference from the mean
es_diff_from_mean = es_abs + x.mean()

out = pd.DataFrame({"ES Absolute": [es_abs], "ES Diff from Mean": [es_diff_from_mean]})
print(out)
```
