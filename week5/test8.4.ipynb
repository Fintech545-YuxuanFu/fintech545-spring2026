```python
import numpy as np
import pandas as pd
from scipy.stats import norm
```

```python
df = pd.read_csv("/Users/fuyuxuan/Downloads/test7_1.csv")
x = df.iloc[:, 0]    # get the only first column

mu = x.mean()
sigma = x.std(ddof=1)

# ES from Normal at 95% confidence level
alpha = 0.05
z = norm.ppf(alpha)
phi = norm.pdf(z)

# ES as an absolute loss
es_abs = - (mu - sigma * (phi / alpha))

# Difference from the mean
es_diff_from_mean = es_abs + mu

out = pd.DataFrame({"ES Absolute": [es_abs], "ES Diff from Mean": [es_diff_from_mean]})
print(out)
```
