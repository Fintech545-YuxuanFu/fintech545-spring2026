```python
import numpy as np
import pandas as pd
from scipy.stats import t
```

```python
df = pd.read_csv("/Users/fuyuxuan/Downloads/test7_2.csv")
x = df.iloc[:, 0]    # get the only first column

alpha = 0.05
nu, loc, scale = t.fit(x)
np.random.seed(0)

# Simulated large sample from fitted t distribution
simulated = t.rvs(df=nu, loc=loc, scale=scale, size=10000000)

# 5% VaR of the simulated sample
var_alpha = np.quantile(simulated, alpha)

# Average values in the left tail
es_left_tail = simulated[simulated <= var_alpha].mean()

# ES as a positive loss
es_abs = - es_left_tail

# Difference from the mean
es_diff_from_mean = es_abs + x.mean()

out = pd.DataFrame({"ES Absolute": [es_abs], "ES Diff from Mean": [es_diff_from_mean]})
print(out)
```
