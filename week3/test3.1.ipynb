```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/Users/fuyuxuan/Downloads/testout_1.3.csv")
A = df.to_numpy(dtype=float)
std = np.sqrt(np.diag(A))
corr = A / np.outer(std, std)

vals, vecs = np.linalg.eigh((corr + corr.T) / 2.0)
vals = np.clip(vals, 0.0, None)

B = vecs @ np.diag(vals) @ vecs.T
s = np.sqrt(np.diag(B))
C_psd = B / np.outer(s, s)

cov_psd = C_psd * np.outer(std, std)
out = pd.DataFrame(cov_psd, columns=df.columns)
print(out)
```
