```python
import pandas as pd
import numpy as np
```

```python
LAMBDA = 0.94
df = pd.read_csv("/Users/fuyuxuan/Downloads/test2.csv")
x = df.to_numpy()
T, N = x.shape

w = np.array([(1 - LAMBDA) * (LAMBDA ** (T - 1 - i)) for i in range(T)], dtype = float)
w = w / w.sum()

mu = (w[:, None] * x).sum(axis=0)
xm = x - mu
cov = xm.T @ (xm * w[:, None])

std = np.sqrt(np.diag(cov))
corr = cov / np.outer(std, std)

out = pd.DataFrame(corr, columns=df.columns)
print(out)
```
