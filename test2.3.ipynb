```python
import pandas as pd
import numpy as np
```

```python
ew_var = 0.97
ew_corr = 0.94

df = pd.read_csv("/Users/fuyuxuan/Downloads/test2.csv")
x = df.to_numpy()
T, N = x.shape

def ew_cov_matrix(x: np.ndarray, lam: float) -> np.ndarray:
    T = x.shape[0]
    w = np.array([(1 - lam) * (lam ** (T - 1 - i)) for i in range(T)], dtype = float)
    w = w / w.sum()
    mu = (w[:, None] * x).sum(axis=0)
    xm = x - mu
    return xm.T @ (xm * w[:, None])

cov97 = ew_cov_matrix(x, ew_var)
var97 = np.diag(cov97)
std97 = np.sqrt(var97)

cov94 = ew_cov_matrix(x, ew_corr)
var94 = np.diag(cov94)
std94 = np.sqrt(var94)
corr94 = cov94 / np.outer(std94, std94)
cov_mix = corr94 * np.outer(std97, std97)

out = pd.DataFrame(cov_mix, columns=df.columns)
print(out)
```
