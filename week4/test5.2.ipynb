```python
import numpy as np
import pandas as pd
```

```python
# Load data
cov_input = pd.read_csv("/Users/fuyuxuan/Downloads/test5_2.csv")
number_of_simulations = 100000

np.random.seed(0)

# Convert it to numpy array
cov_matrix = cov_input.values

# Zero mean vector
mu = np.zeros(cov_matrix.shape[0])

# Simulate multivariate normal draws
simulated_data = np.random.multivariate_normal(mu, cov_matrix, number_of_simulations)

# Compute output covariance matrix
cov_output = np.cov(simulated_data, rowvar=False, ddof=0)

# Convert it to DataFrame
cov_output_df = pd.DataFrame(cov_output, columns=cov_input.columns, index=cov_input.columns)

print(cov_output_df)
```
