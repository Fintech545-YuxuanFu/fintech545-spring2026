```python
import numpy as np
import pandas as pd
from scipy.stats import t
```

```python
df = pd.read_csv("/Users/fuyuxuan/Downloads/test7_2.csv")
x = df.iloc[:, 0].to_numpy()

# Fit Student-t distribution using MLE (df, loc, scale)
df_hat, loc_hat, scale_hat = t.fit(x)

# Simulate returns from fitted t distribution
np.random.seed(0)
number_of_simulation = 100000
sim = t.rvs(df_hat, loc=loc_hat, scale=scale_hat, size=number_of_simulation)

# Calculate mean and 5% quantile under fitted t distribution
alpha = 0.05
mu_sim = sim.mean()
quantile_005 = np.quantile(sim, alpha)

# Convert to VaR outputs
var_abs = -quantile_005
var_diff = mu_sim - quantile_005

output = pd.DataFrame([{
    "VaR Absolute": var_abs,
    "VaR Diff from Mean": var_diff
}])
print(output)
```
