```python
import numpy as np
import pandas as pd
from scipy.stats import norm
```

```python
df = pd.read_csv("/Users/fuyuxuan/Downloads/test7_1.csv")
x = df.iloc[:, 0].to_numpy()

# Calculate the sample mean and std (ddof=1)
mu = x.mean()
sigma = x.std(ddof=1)

# Calculate 5% quantile under Normal(mu, sigma)
alpha = 0.05
quantile_005 = mu + norm.ppf(alpha) * sigma

# Convert to VaR outputs
var_abs = -quantile_005
var_diff = mu - quantile_005

output = pd.DataFrame([{
    "VaR Absolute": var_abs,
    "VaR Diff from Mean": var_diff
}])
print(output)
```
