```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/Users/fuyuxuan/Downloads/testout_3.1.csv")
A = df.to_numpy(dtype=float)
A = (A + A.T) / 2.0

n = A.shape[0]
L = np.zeros_like(A)

for i in range(n):
    s = 0.0
    for k in range(i):
        s += L[i, k] * L[i, k]

    pivot = A[i, i] - s
    if pivot <= 0.0:
        L[i, i] = 0.0
    else:
        L[i, i] = np.sqrt(pivot)

    for j in range(i + 1, n):
        s2 = 0.0
        for k in range(i):
            s2 += L[j, k] * L[i, k]

        if L[i, i] > 0.0:
            L[j, i] = (A[j, i] - s2) / L[i, i]
        else:
            L[j, i] = 0.0

out = pd.DataFrame(L, columns=df.columns)
print(out)
```
